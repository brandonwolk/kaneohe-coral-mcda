```python
# [CELL: DOWNLOAD_CWB_WATER_QUALITY]
import os
import requests

raw_dir = '/content/drive/MyDrive/Capstone/01 Raw Data'
os.makedirs(raw_dir, exist_ok=True)

cwb_url = "https://pae-paha.pacioos.hawaii.edu/erddap/tabledap/cwb_water_quality.csv?time%2Clongitude%2Clatitude%2Clocation_id%2Clocation_name%2Centerococcus%2Cclostridium%2Cwater_temperature%2Csalinity%2Cturbidity%2Cph%2Coxygen%2Coxygen_saturation&time%3E=2000-01-01T00%3A00%3A00Z&time%3C=2025-12-31T23%3A59%3A59Z&longitude%3E=-157.95&longitude%3C=-157.70&latitude%3E=21.35&latitude%3C=21.65"
cwb_path = os.path.join(raw_dir, 'cwb_water_quality.csv')

print("Downloading CWB Water Quality data...")
r_cwb = requests.get(cwb_url)
with open(cwb_path, 'wb') as f:
    f.write(r_cwb.content)
print(f"Success! Saved to: {cwb_path}")
```
